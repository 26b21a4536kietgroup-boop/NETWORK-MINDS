```python
import numpy as np
import sympy as sp

# ============================================================
# NETWORK INFLUENCE PROPAGATION (C8)
# Matrix Powers using Cayley-Hamilton + Consensus Dynamics
# ============================================================

# ---------- INPUT ----------
n_people = 5

# Influence matrix A
# Each row represents the weights used to calculate
# the next opinion of that person.
A = sp.Matrix(
    [
        [0.50, 0.20, 0.10, 0.10, 0.10],
        [0.20, 0.50, 0.10, 0.10, 0.10],
        [0.10, 0.10, 0.50, 0.20, 0.10],
        [0.10, 0.10, 0.20, 0.50, 0.10],
        [0.10, 0.10, 0.10, 0.10, 0.60],
    ]
)

# Initial opinions of P1, P2, P3, P4, P5
x0 = sp.Matrix([1.0, 0.5, -1.0, 0.0, 0.8])

# Number of discussion rounds
rounds = 10

# Consensus tolerance
epsilon = 0.001


# ============================================================
# 1. DISPLAY INPUT
# ============================================================

print("=" * 65)
print("       NETWORK INFLUENCE PROPAGATION - C8 PROJECT")
print("=" * 65)

print("\nInfluence Matrix A:")
sp.pprint(A)

print("\nInitial Opinion Vector x0:")
sp.pprint(x0)

print("\nNumber of Discussion Rounds:", rounds)


# ============================================================
# 2. CHARACTERISTIC POLYNOMIAL
# ============================================================

lambda_symbol = sp.symbols("lambda")

char_poly = A.charpoly(lambda_symbol).as_expr()

print("\n" + "=" * 65)
print("CHARACTERISTIC POLYNOMIAL")
print("=" * 65)

print("p(lambda) =")
sp.pprint(char_poly)


# ============================================================
# 3. CAYLEY-HAMILTON THEOREM
# ============================================================

print("\n" + "=" * 65)
print("CAYLEY-HAMILTON THEOREM")
print("=" * 65)

print("\nBy Cayley-Hamilton:")
print("p(A) = 0")

coefficients = sp.Poly(char_poly, lambda_symbol).all_coeffs()

print("\nCharacteristic polynomial coefficients:")
print(coefficients)


# ============================================================
# 4. CALCULATE A^n USING CAYLEY-HAMILTON
# ============================================================

# SymPy's rem() reduces lambda^n modulo the characteristic
# polynomial. Therefore A^n can be represented as a combination
# of I, A, A^2, A^3 and A^4.

poly_n = sp.rem(
    sp.Poly(lambda_symbol**rounds, lambda_symbol),
    sp.Poly(char_poly, lambda_symbol),
)

print("\n" + "=" * 65)
print("CAYLEY-HAMILTON REDUCTION")
print("=" * 65)

print(f"\nA^{rounds} can be reduced using:")
sp.pprint(poly_n.as_expr())

# Convert the reduced polynomial into a matrix
A_n_CH = sp.zeros(n_people)

poly_expression = sp.Poly(poly_n, lambda_symbol)

for (power,), coefficient in poly_expression.terms():
    A_n_CH += coefficient * (A**power)

A_n_CH = sp.simplify(A_n_CH)

print(f"\nMatrix A^{rounds} using Cayley-Hamilton:")
sp.pprint(A_n_CH)


# ============================================================
# 5. FINAL OPINIONS
# ============================================================

x_n = sp.simplify(A_n_CH * x0)

print("\n" + "=" * 65)
print(f"FINAL OPINIONS AFTER {rounds} ROUNDS")
print("=" * 65)

for i in range(n_people):
    print(f"P{i+1} = {float(x_n[i]):.6f}")


# ============================================================
# 6. OPINION PROPAGATION ROUND BY ROUND
# ============================================================

print("\n" + "=" * 65)
print("OPINION PROPAGATION")
print("=" * 65)

current = x0

print("\nRound 0:")
print([round(float(v), 4) for v in current])

for r in range(1, rounds + 1):
    current = A * current

    print(f"Round {r}:")
    print([round(float(v), 4) for v in current])


# ============================================================
# 7. CONSENSUS CHECK
# ============================================================

final_values = np.array([float(v) for v in x_n])

maximum = np.max(final_values)
minimum = np.min(final_values)

difference = maximum - minimum

print("\n" + "=" * 65)
print("CONSENSUS ANALYSIS")
print("=" * 65)

print(f"\nMaximum opinion = {maximum:.6f}")
print(f"Minimum opinion = {minimum:.6f}")
print(f"Difference      = {difference:.6f}")

if difference < epsilon:
    print("\nRESULT: CONSENSUS REACHED")
    print("All people have approximately the same opinion.")
else:
    print("\nRESULT: CONSENSUS NOT REACHED")
    print("People still have different opinions.")


# ============================================================
#
```
